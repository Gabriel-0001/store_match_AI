```python
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
ROOT = Path.cwd().parent

df_main = pd.read_csv(ROOT/"data/raw/osm_mcdonalds.csv")
df_main.head()
```

```python
df_main.dtypes
```

```python
df_main['id'] = df_main['id'].astype(str)
```

```python
df_main[df_main['addr:street'].isna()]
```

```python
df_main = df_main.dropna(subset=['addr:street', 'addr:city', 'addr:postcode', 'addr:housenumber'])
```

```python
df_main
```

```python
df_main = df_main.drop(columns=['element', 'phone', 'website', 'geometry'])
```

```python
df_main
```

```python
df_main.isna().sum()
```

```python
df_main = df_main.rename(columns={
    'addr:street': 'street',
    'addr:city': 'city',
    'addr:postcode': 'postcode',
    'addr:housenumber': 'housenumber'
})
```

```python
df_main.to_csv(ROOT/"data/processed/osm_mcdonalds.csv", index=False)
```
